```python
import requests, json

# Si estás dentro del mismo host/compose y compartís red:
OLLAMA_URL = "http://ollama:11434"   # o "http://localhost:11434" si pruebas desde el host

# Verifica si el endpoint responde
r = requests.get(f"{OLLAMA_URL}/api/tags")
print(r.status_code)
print(r.json())

# 4.1 Ver modelos instalados
resp = requests.get(f"{OLLAMA_URL}/api/tags", timeout=30)
print("Modelos instalados:", [m["name"] for m in resp.json().get("models", [])])

# 4.2 Generar un texto mínimo con un modelo (ajusta el nombre si es distinto)
payload = {
    "model": "llama3.1:8b-instruct-fp16",            # cámbialo a un modelo que tengas (p.ej. "mistral:latest")
    "prompt": "Responde con 'OK'.",
    "stream": False
}
resp = requests.post(f"{OLLAMA_URL}/api/generate", json=payload, timeout=600)
print("Respuesta del modelo:", resp.json().get("response"))
```

```python
import requests

OLLAMA_URL = "http://ollama:11434"   # dentro del contenedor jupyter

prompt = "Escribe 'OK' y nada más."

payload = {
    "model": "llama3.1:8b-instruct-fp16",
    "prompt": prompt,
    "stream": False
}

resp = requests.post(f"{OLLAMA_URL}/api/generate", json=payload, timeout=120)
print("Status:", resp.status_code)
print("JSON:", resp.json())
```
