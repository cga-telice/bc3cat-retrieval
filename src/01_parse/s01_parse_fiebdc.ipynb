```python
"""
FIEBDC-3-2016-rev2023 Parser

This script parses a FIEBDC-3-2016-rev2023 file into JSON format.
It extracts parameter and text-related information from the input file.

Reference document: https://www.fiebdc.es/fiebdc-32016-2/
"""

import os
from pathlib import Path
import string
import json
import re
import csv
from io import StringIO
from math import sin, cos, tan, asin, acos, atan, atan2, sqrt, fabs as ABS, floor as INT

# Constants
PARAMETER_RANGE = ("A", "B", "C", "D", "F", "G", "H", "I", "J", "K")
ALL_LETTERS = string.ascii_lowercase + string.ascii_uppercase

# Formula processing functions (previously in z_formula_processing.py)
def quote_second_term(text):
    """Quote single characters (parameter labels)."""
    pattern = r'(\w+\s*([=<>!]=|[=<>])\s*)(\w+)'
    return re.sub(pattern, lambda m: f'{m.group(1)}"{m.group(3)}"', text)

def translate_formula_to_python(formula):
    """Translate the formula string into Python syntax."""
    python_equivalents = {
        "@": " or ", "&": " and ", "^": "**", "=": "==", "<>": "!=",
        "ABS": "ABS", "INT": "INT", "ROUND": "round",
        "SIN": "sin", "COS": "cos", "TAN": "tan",
        "ASIN": "asin", "ACOS": "acos", "ATAN": "atan", "ATAN2": "atan2",
        "SQRT": "sqrt", "ATOF": "float", "FTOA": "str",
    }
    
    for custom, py_equiv in python_equivalents.items():
        formula = re.sub(re.escape(custom), py_equiv, formula)
    
    formula = re.sub(r'\(%(\w)=(\w)\)', r'(1 if \1=="\2" else 0)', formula)
    formula = quote_second_term(formula)
    
    return formula

def setup_paths(raw_file):
    """Set up file paths for input and output files."""
    raw_dir = Path("raw")
    intermediate_dir = Path("intermediate")
    file_path_raw = raw_dir / raw_file
    file_path_intermediate = intermediate_dir / raw_file.replace('.txt', '.json')
    return file_path_raw, file_path_intermediate

def extract_parameters(p_paragraph_text):
    """Extract parameters from a paragraph."""
    parameter_data = {}
    parameter_label = iter(PARAMETER_RANGE)

    for item in p_paragraph_text:
        params_and_values = [item for item in item.split('|')[-1].split('\\') if item]

        # Skip invalid entries like "#Parámetros"
        if params_and_values[0].startswith("#"):
            continue
        
        labeled_values = [{"label": ALL_LETTERS[k], "value": value} for k, value in enumerate(params_and_values[1:])]
        try:
            key = next(parameter_label)
        except:
            print(p_paragraph_text)
        parameter_data[key] = {
            "label": params_and_values[0],
            "values": labeled_values
        }
    
    return parameter_data

def parse_text_items(text_formula):
    """
    Parse text formula into a list of string items.
    Handles:
    1. Python tuples of strings
    2. Comma-separated quoted strings
    3. Raw formulas and other text formats
    """
    # If input is already a tuple, convert directly to list
    if isinstance(text_formula, tuple):
        return list(text_formula)
    
    # Ensure we're working with a string
    if not isinstance(text_formula, str):
        return []
        
    # Strip any outer quotes if present
    text_formula = text_formula.strip()
    if text_formula.startswith('"') and text_formula.endswith('"'):
        text_formula = text_formula[1:-1]
    
    # Initialize variables
    items = []
    current_item = ""
    i = 0
    
    while i < len(text_formula):
        char = text_formula[i]
        
        if char == '"':
            # Skip this quote and look for the ending quote
            i += 1
            while i < len(text_formula):
                if text_formula[i] == '"':
                    # Found potential ending quote
                    if i + 1 < len(text_formula) and text_formula[i + 1] == '"':
                        # This is an escaped quote, include it and skip the next quote
                        current_item += '"'
                        i += 2
                    else:
                        # This is an ending quote
                        items.append(current_item)
                        current_item = ""
                        # Skip until next starting quote or end
                        i += 1
                        while i < len(text_formula) and text_formula[i] != '"':
                            if not text_formula[i].isspace() and text_formula[i] != ',':
                                # Found non-space, non-comma character - treat as raw text
                                return [text_formula]
                            i += 1
                        break
                else:
                    current_item += text_formula[i]
                    i += 1
        else:
            # If we find non-space character outside of quotes, 
            # treat as a regular text formula
            if not char.isspace() and char != ',':
                return [text_formula]
            i += 1
    
    # Handle any remaining item
    if current_item:
        items.append(current_item)
        
    return items

def extract_text_variable(text_variable):
    """
    Extract text variables and always return two values:
    - name: the variable name
    - formula: a list (or nested list) of formulas linked to the text variable.
    """
    import re

    # Clean up the input
    if isinstance(text_variable, list):
        text_variable = ' '.join(text_variable)

    # Remove any comments (everything after #)
    text_variable = re.sub(r'#.*$', '', text_variable, flags=re.MULTILINE).strip()


    # Flatten the variable
    def refine_flattening(text_variable):
        text_variable = re.sub(r'#.*$', '', text_variable, flags=re.MULTILINE).strip()
        text_variable = re.sub(r',\s*\n', ',', text_variable)  # Handle trailing commas on newlines
        text_variable = re.sub(r'\s*\n', ' ', text_variable)  # Flatten remaining newlines
        return text_variable

    flattened_variable = refine_flattening(text_variable)

    # Match variable structure
    variable_match = re.match(r'^\$(\w+)\s*(?:\(([\d\s,]+)\))?\s*=\s*(.*)$', flattened_variable.strip())
    if not variable_match:
        # Treat as a formula for concatenations or operations
        if flattened_variable.strip().startswith('$'):
            name = flattened_variable.split('=')[0].strip()
            formula = flattened_variable.split('=')[1].strip()
            return name, formula  # Treat the entire right-hand side as the formula
        return None, "Structure not matched with refined regex"

    # Extract dimensions and content
    name, dimensions, formula = variable_match.groups()
    dimensions = tuple(map(int, re.split(r'\s*,\s*', dimensions))) if dimensions else None

    # Split formulas while respecting quoted strings
    formulas = [v.strip() for v in re.split(r',(?=(?:[^"]*"[^"]*")*[^"]*$)', formula)]

        # Handle direct text concatenation
    if '+' in formula:
        full_formula = re.sub(r'\s*\+\s*', ' + ', formula)  # Normalize spaces around "+"
        return name, translate_formula_to_python(full_formula)

    # Handle quoted text (string or array of strings)
    if formula.startswith('"'):
        items = []
        current_item = ""
        in_quotes = False
        i = 0

        while i < len(formula):
            char = formula[i]

            if char == '"':
                if not in_quotes:
                    in_quotes = True
                else:
                    if i + 1 < len(formula) and formula[i + 1] == '"':
                        current_item += '"'
                        i += 1
                    else:
                        in_quotes = False
                        items.append(current_item)
                        current_item = ""
                        while i + 1 < len(formula) and formula[i + 1] not in ('"', ','):
                            i += 1
            elif in_quotes:
                current_item += char
            elif not char.isspace() and char != ',':
                if not items:
                    return name, translate_formula_to_python(formula)
            i += 1

        if current_item:
            items.append(current_item)

    # Translate formulas to Python
    values = [translate_formula_to_python(f) for f in formulas]

    # Handle dimensions
    if dimensions:
        # Parse structured data into a nested list
        def reshape(values, dims):
            if len(dims) == 1:
                return values[:dims[0]]
            size = dims[0]
            step = len(values) // size
            return [reshape(values[i * step:(i + 1) * step], dims[1:]) for i in range(size)]

        reshaped_values = reshape(values, dimensions)
        return name, reshaped_values  # Return the name and nested list of formulas

    else:
        # Treat as a formula for concatenations or operations
        return name, values

def reshape_list(flat_list, dims):
    """Reshape a flat list into nested lists according to dims."""
    if len(dims) == 1:
        return flat_list
    size = dims[0]
    sub_dims = dims[1:]
    sub_size = 1
    for d in sub_dims:
        sub_size *= d
    return [reshape_list(flat_list[i * sub_size:(i + 1) * sub_size], sub_dims) 
            for i in range(size)]

def extract_text(t_paragraph_text):
    """Extract and combine text from a paragraph."""
    combined_text = " ".join(t_paragraph_text)
    # Remove TEXTO marker and ending marker \|
    combined_text = combined_text.replace('\\TEXTO\\', "")
    combined_text = combined_text.replace('\\|', "")
    return combined_text.strip()

class FIEBDCParser:
    def __init__(self):
        self.data = {}
        self.current_id = None
        self.p_paragraph = False
        self.p_paragraph_text = []
        self.t_paragraph = False
        self.t_paragraph_text = []
        self.current_variable = None
        self.current_formulas = []

    def process_line(self, line):
        """Process a single line of the input file."""
        #print(f"DEBUG: Processing line: {line}")
        line = line.strip()

        if not line:  # Skip empty lines
            #print("DEBUG: Skipping empty line")
            return

        # Add this debug print for RESUMEN and TEXTO lines
        if '\\RESUMEN\\' in line or '\\TEXTO\\' in line:
            #print("\nDEBUG Found special line:", line)
            pass

        # Simple string-based normalization
        if '\\' in line and 'RESUMEN' in line:
            parts = line.split('RESUMEN')
            line = '\\RESUMEN\\' + parts[1]
        if '\\' in line and 'TEXTO' in line:
            parts = line.split('TEXTO')
            line = '\\TEXTO\\' + parts[1]

        # Handle RESUMEN and TEXTO lines first
        if '\\RESUMEN\\' in line:
            self._process_resumen(line)
            return
        elif '\\TEXTO\\' in line:
            #print(f"DEBUG: Detected start of TEXTO block: {line}")
            self.t_paragraph = True #changed
            self.t_paragraph_text.append(line) #changed
            #self._process_texto(line)
            return
        
        if line.startswith("#") or line.startswith("%"):
            self._finalize_paragraphs()
            self._finalize_text_variable()
        elif line.startswith("~C"):
            self._finalize_paragraphs()
            self._finalize_text_variable()
            self._process_concept(line)
        elif line.startswith("~P"):
            self._finalize_paragraphs()
            self._finalize_text_variable()
            self._process_parameter_paragraph(line)
        elif self.p_paragraph and line.startswith("\\"):
            self.p_paragraph_text.append(line)
        elif self.t_paragraph:
            if line.endswith("\\|"):  # End of TEXTO block
                self.t_paragraph_text.append(line)  # Add the final line
                #print(f"DEBUG: Finalizing TEXTO block: {self.t_paragraph_text}")
                self._finalize_paragraphs()  # Finalize the paragraph
            else:
                self.t_paragraph_text.append(line)  # Add the line to TEXTO content
                #print(f"DEBUG: Appending to TEXTO block: {line}")
                
        elif line.startswith("$"):
            self._finalize_paragraphs()
            self._finalize_text_variable()
            # This is a new variable definition
            match = re.match(r'^\$(\w+)(?:\(([\d\s,]+)\))?\s*=\s*(.*)$', line)
            if match:
                name, dims, initial_formula = match.groups()
                self.current_variable = f"${name}({dims}) = " if dims else f"${name} = "
                #print(f"DEBUG: Detected new variable: {self.current_variable}")  # Add debug log
                if initial_formula:
                    self.current_formulas.append(initial_formula.rstrip(','))
                    #print(f"DEBUG: Initial formula: {initial_formula.rstrip(',')}")  # Add debug log
                if not (line.endswith(',') or line.endswith('=')):
                    # Single line definition
                    self._finalize_text_variable()
        elif self.current_variable:
            # Continue collecting text variable lines
            clean_line = re.sub(r'#.*$', '', line).rstrip()
            if clean_line:
                if clean_line.endswith(','):
                    #print(f"DEBUG: Appending to current variable: {clean_line}")  # Add debug log
                    self.current_formulas.append(clean_line.rstrip(','))
                else:
                    #print(f"DEBUG: Final line for current variable: {clean_line}")  # Add debug log
                    self.current_formulas.append(clean_line)
                    self._finalize_text_variable()

    def _finalize_text_variable(self):
        """Process any pending text variable definition."""
        if self.current_variable:
            complete_text = self.current_variable + " " + ", ".join(self.current_formulas)
            #print(f"DEBUG: Finalizing variable: {complete_text}")  # Add debug log
            self._process_text_variable(complete_text)
            self.current_variable = None
            self.current_formulas = []

    def _finalize_text_variable(self):
        """Process any pending text variable definition."""
        
        if self.current_variable:
            # Extract the initial definition parts
            match = re.match(r'^\$(\w+)(?:\(([\d\s,]+)\))?\s*=\s*(.*)$', self.current_variable)
            if match:
                name, dim_str, initial_formula = match.groups()
                initial_formula = initial_formula.strip().rstrip(',')
            
                # Combine all formulas
                all_formulas = [initial_formula] if initial_formula else []
                all_formulas.extend(self.current_formulas)
            
                # Reconstruct the variable definition with all formulas
                if dim_str:
                    complete_text = f"${name}({dim_str}) = " + ", ".join(all_formulas)
                else:
                    complete_text = f"${name} = " + ", ".join(all_formulas)
                
                # Debug logging for finalization
                #print(f"DEBUG: Finalizing variable: {complete_text}")
                
                self._process_text_variable(complete_text)
        
            self.current_variable = None
            self.current_formulas = []

    def _finalize_paragraphs(self):
        """Finalize processing of paragraphs."""
        if self.t_paragraph:
            #print(f"DEBUG: Finalizing TEXTO paragraph: {self.t_paragraph_text}")
            # Join lines, remove markers (\TEXTO\ and \|), and strip whitespace
            full_text = " ".join(self.t_paragraph_text).replace("\\TEXTO\\", "").replace("\\|", "").strip()
            self.data[self.current_id]['texto'] = full_text
            self.t_paragraph_text = []
            #print(f"DEBUG: TEXTO block saved: {full_text}")
        elif self.p_paragraph:
            self.data[self.current_id]['parameters'] = extract_parameters(self.p_paragraph_text)
            self.p_paragraph_text = []
        self.p_paragraph = False
        self.t_paragraph = False
    
    def _process_concept(self, line):
        """Process a concept line."""
        parts = line.split('|')
        self.current_id = parts[1].strip()
        self.data[self.current_id] = {
            'ud': parts[2].strip(),
            'concept': parts[3].strip(),
            'text_variables': {},
            'resumen': [],
            'texto': []
        }

    def _process_parameter_paragraph(self, line):
        """Process a parameter paragraph."""
        self._finalize_paragraphs()
        parts = line.split('|')
        self.current_id = parts[1].strip()
        self.p_paragraph_text = [line]
        self.p_paragraph = True

    def _process_text_variable(self, line):
        """Process a text variable line."""
        #print(f"DEBUG: Processing text variable: {line}")
        key, formula = extract_text_variable(line)
        if key is not None:  # Only process if we successfully parsed the variable
            self.data[self.current_id]['text_variables'][key] = formula

    def _process_resumen(self, line):
        """Process a RESUMEN paragraph."""
        #print("DEBUG RESUMEN line:", line)
        if '\\RESUMEN\\' in line:
            try:
                content = line.split('\\RESUMEN\\')[1].replace('\\|', '').strip()
                # Remove leading backslash if present
                if content.startswith('\\'):
                    content = content[1:]
                self.data[self.current_id]['resumen'] = content
                #print("DEBUG RESUMEN saved:", self.data[self.current_id]['resumen'])
            except IndexError:
                #print("DEBUG RESUMEN error: couldn't parse line:", line)
                pass
        else:
            # Handle case with spaces
            try:
                parts = [p.strip() for p in line.split('\\') if p.strip()]
                if 'RESUMEN' in parts:
                    idx = parts.index('RESUMEN')
                    if idx + 1 < len(parts):
                        content = parts[idx + 1].replace('|', '').strip()
                        if content.startswith('\\'):
                            content = content[1:]
                        self.data[self.current_id]['resumen'] = content
                        #print("DEBUG RESUMEN saved (alternative):", content)
            except Exception as e:
                #print("DEBUG RESUMEN error:", str(e), line)
                pass

    def _process_texto(self, line):
        """Process a TEXTO paragraph."""
        #print("DEBUG TEXTO line:", line)
        if '\\TEXTO\\' in line:
            try:
                content = line.split('\\TEXTO\\')[1].replace('\\|', '').strip()
                # Remove leading backslash if present
                if content.startswith('\\'):
                    content = content[1:]
                self.data[self.current_id]['texto'] = content
                #print("DEBUG TEXTO saved:", self.data[self.current_id]['texto'])
            except IndexError:
                #print("DEBUG TEXTO error: couldn't parse line:", line)
                pass
        else:
            # Handle case with spaces
            try:
                parts = [p.strip() for p in line.split('\\') if p.strip()]
                if 'TEXTO' in parts:
                    idx = parts.index('TEXTO')
                    if idx + 1 < len(parts):
                        content = parts[idx + 1].replace('|', '').strip()
                        if content.startswith('\\'):
                            content = content[1:]
                        self.data[self.current_id]['texto'] = content
                        print("DEBUG TEXTO saved (alternative):", content)
            except Exception as e:
                #print("DEBUG TEXTO error:", str(e), line)
                pass

def parse_fiebdc_file(file_path):
    """Parse the FIEBDC file and return the processed data."""
    parser = FIEBDCParser()
   
    with open(file_path, 'r', encoding='UTF-8') as file: 
        for line in file:
            parser.process_line(line)

    parser._finalize_paragraphs()  # Process any remaining paragraphs
    return parser.data

def main(raw_file):
    os.chdir('/work/data')  # Running in a container
    file_path_raw, file_path_intermediate = setup_paths(raw_file)

    processed_data = parse_fiebdc_file(file_path_raw)

    with open(file_path_intermediate, 'w', encoding='UTF-8') as json_file:
        json.dump(processed_data, json_file, indent=4, ensure_ascii=False)

    print(f"Processed data has been saved to {file_path_intermediate}")
```

```python
main("BPA_2024_v2_OEB_mod_utf8.txt")
```
