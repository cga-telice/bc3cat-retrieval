```python
documents_path = '/work/data/processed/OEB_resumen.pkl'
documents = load_documents(documents_path)
# Convert documents to serializable dictionaries
def document_to_dict(doc):
    return {
        "id": doc.id_,
        "item_key": doc.metadata.get("item_key"),
        "parent_key": doc.metadata.get("parent_key"),
        "ud": doc.metadata.get("ud"),
        "concept": doc.metadata.get("concept"),
        "parameters": doc.metadata.get("parameters"),
        "text": doc.text,
    }

serializable_docs = [document_to_dict(doc) for doc in documents]

# Save to JSON file
output_file = "OEB_resumen.json"
with open(output_file, "w", encoding="utf-8") as f:
    json.dump(serializable_docs, f, ensure_ascii=False, indent=2)

print(f"Saved {len(serializable_docs)} documents to {output_file}")
```
