```python
# Add project root to path
import sys
from pathlib import Path
project_root = Path('/work')
sys.path.append(str(project_root))

# Import the quick_builder functions
from sandbox.splade.indexing.quick_builder import (
    quick_baseline_setup,
    load_baseline_index,
    create_baseline_index,
    test_search_functionality,
    batch_search_example,
    load_documents_from_pickle
)

print("✅ Quick builder imports successful!")
```

```python
# Quick setup with default parameters
print("🚀 Creating baseline index...")

try:
    index = quick_baseline_setup(
        file_base="OEB",
        vocab_path="/work/sandbox/splade/vocabularies",
        data_path="/work/data/processed",
        use_filtered=True,
        max_features=5000,  # Smaller for testing
        save_index=True
    )
    print("✅ Baseline index created successfully!")
    
except Exception as e:
    print(f"❌ Error creating index: {e}")
    import traceback
    traceback.print_exc()
```

```python
# Load the existing baseline index
print("📁 Loading existing baseline index...")

try:
    index = load_baseline_index(
        file_base="OEB",
        vocab_path="/work/sandbox/splade/vocabularies",
        index_path="/work/sandbox/splade/baseline_indices",
        use_filtered=True,
        max_features=5000  # Use same size as when created
    )
    print("✅ Baseline index loaded successfully!")
    
    # Quick verification
    stats = index.get_index_statistics()
    print(f"📊 Loaded index with {stats['total_documents']:,} documents")
    
except Exception as e:
    print(f"❌ Error loading index: {e}")
    print("💡 If index doesn't exist, you may need to create it first with:")
    print("   index = quick_baseline_setup(file_base='OEB', max_features=10000)")
    import traceback
    traceback.print_exc()
```

```python
# Test single query
test_query = "canalización tubos polietileno"
print(f"🔍 Testing query: '{test_query}'")

try:
    results = index.search(test_query, top_k=5, min_score=0.01)
    
    print(f"Found {len(results)} results:")
    for i, result in enumerate(results):
        print(f"\n{i+1}. Item Key: {result.item_key}")
        print(f"   Score: {result.score:.4f}")
        print(f"   Text preview: {result.text[:150]}...")
        if result.metadata:
            print(f"   Metadata keys: {list(result.metadata.keys())}")
            
except Exception as e:
    print(f"❌ Search error: {e}")
```
